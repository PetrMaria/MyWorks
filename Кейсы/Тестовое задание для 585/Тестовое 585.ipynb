```python
import pandas as pd
```

Загрузим данные из файла

```python
df = pd.read_excel('585.xlsx')
```

Проверим на актуальность типа данных

```python
df.info()
```

```python
df.head()
```

Для начала отфильтруем данные по заданному периоду: с 1.04.23 по 10.04.23


```python
df = df[(df['ДатаПродажи'] >= '2023-04-01') & (df['ДатаПродажи'] <= '2023-04-10')]
```

Оставим только данные изделий с общей суммой и себестоимостью больше 0 рублей

```python
df = df[(df['ОбщаяСумма'] > 0) & (df['Себестоимость'] > 0)]
```

Вычислим наценку на ювелирные изделия и добавим столбец в датафрейм

```python
df['Наценка'] = ((df['ОбщаяСумма'] - df['Себестоимость']) / df['Себестоимость']) * 100
```

```python
df.head()
```

Сгруппируем данные об изделиях по металлу

```python
df_silver = df[df['ТоварнаяГруппа'].str.startswith('СИ')]
df_gold = df[~df['ТоварнаяГруппа'].str.startswith('СИ')]
```

```python
df_silver.head()
```

Посчитаем среднюю наценку для изделий из серебра

```python
df_silver['Наценка'].mean()
```

```python
df_gold.head()
```

Посчитаем среднюю наценку для изделий из золота

```python
df_gold['Наценка'].mean()
```
